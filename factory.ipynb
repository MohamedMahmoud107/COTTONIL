```python
import pandas as pd

df = pd.read_csv(r"D:\HVIA\New folder\Cottonil_project\COTTONIL\data\factories.csv", encoding="utf-8-sig")

df.head()
```

```python
from egygeo import get_coordinates

df["governorate_ar"] = df["governorate_ar"].str.strip()

df[["latitude", "longitude"]] = df["governorate_ar"].apply(
    lambda name: pd.Series(get_coordinates(name))
)

df[["factory_id", "factory_name_ar", "governorate_ar",
    "latitude", "longitude"]]
```

```python
import folium
from IPython.display import display

# Center the map on Egypt
m = folium.Map(
    location=[26.8, 30.8],
    zoom_start=6,
    tiles="OpenStreetMap"
)

# Add one marker for each factory
for _, row in df.iterrows():

    if pd.isna(row["latitude"]) or pd.isna(row["longitude"]):
        continue

    popup_html = f"""
    <div style="font-family: Arial; direction: rtl; text-align: right;">
        <b>اسم المصنع:</b> {row['factory_name_ar']}<br>
        <b>كود المصنع:</b> {row['factory_id']}<br>
        <b>المحافظة:</b> {row['governorate_ar']}<br>
        <b>نوع الإنتاج:</b> {row['production_focus']}
    </div>
    """

    folium.Marker(
        location=[row["latitude"], row["longitude"]],
        tooltip=row["factory_name_ar"],
        popup=folium.Popup(popup_html, max_width=300),
        icon=folium.Icon(color="red", icon="industry", prefix="fa")
    ).add_to(m)

display(m)
```
